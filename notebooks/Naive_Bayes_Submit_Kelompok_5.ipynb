## Tahap 2 — Hitung Manual (15 menit)

**Kasus uji:** Hari ini cuaca **Sunny**, suhu **Cool**, kelembapan **High**, angin **Strong**.  
**Pertanyaan:** Apakah cocok bermain tenis (`Play = Yes` atau `No`)?

Rumus (slide 2.2):
$$P(K \mid F_1, F_2, \dots, F_n) \propto P(K) \times \prod_{i=1}^{n} P(F_i \mid K)$$

Kita akan menghitung *tanpa* Laplace smoothing terlebih dahulu (sesuai rumus di slide).

```python
test = {"Outlook": "Sunny", "Temperature": "Cool", "Humidity": "High", "Wind": "Strong"}
test
```

### 2.1 Probabilitas Prior P(K)

```python
# TODO 2: Hitung P(Yes) dan P(No)
n_total = len(df)                          # total baris
n_yes = (df['Play'] == 'Yes').sum()        # jumlah Play == 'Yes'
n_no = (df['Play'] == 'No').sum()          # jumlah Play == 'No'

p_yes = n_yes / n_total
p_no = n_no / n_total

print(f"P(Yes) = {p_yes:.4f}")
print(f"P(No)  = {p_no:.4f}")
```

### 2.2 Probabilitas Kondisional P(F_i | K)

Untuk setiap fitur dalam kasus uji, hitung berapa kali nilainya muncul dalam tiap kelas, dibagi total baris kelas tersebut.

```python
df_yes = df[df["Play"] == "Yes"]
df_no = df[df["Play"] == "No"]

# TODO 3: Hitung 4 probabilitas kondisional untuk kelas Yes
p_sunny_yes  = (df_yes['Outlook'] == 'Sunny').sum() / len(df_yes)
p_cool_yes   = (df_yes['Temperature'] == 'Cool').sum() / len(df_yes)
p_high_yes   = (df_yes['Humidity'] == 'High').sum() / len(df_yes)
p_strong_yes = (df_yes['Wind'] == 'Strong').sum() / len(df_yes)

# TODO 4: Hitung 4 probabilitas kondisional untuk kelas No
p_sunny_no  = (df_no['Outlook'] == 'Sunny').sum() / len(df_no)
p_cool_no   = (df_no['Temperature'] == 'Cool').sum() / len(df_no)
p_high_no   = (df_no['Humidity'] == 'High').sum() / len(df_no)
p_strong_no = (df_no['Wind'] == 'Strong').sum() / len(df_no)

print("Kelas Yes:", p_sunny_yes, p_cool_yes, p_high_yes, p_strong_yes)
print("Kelas No :", p_sunny_no,  p_cool_no,  p_high_no,  p_strong_no)
```

### 2.3 Posterior & Keputusan

```python
# TODO 5: Hitung skor posterior (proporsional, belum dinormalisasi)
score_yes = p_yes * p_sunny_yes * p_cool_yes * p_high_yes * p_strong_yes
score_no  = p_no  * p_sunny_no  * p_cool_no  * p_high_no  * p_strong_no

# TODO 6: Normalisasi sehingga total = 1
total = score_yes + score_no
post_yes = score_yes / total
post_no  = score_no / total

prediksi_manual = "Yes" if post_yes > post_no else "No"

print(f"P(Yes | x) = {post_yes:.4f}")
print(f"P(No  | x) = {post_no:.4f}")
print(f"Prediksi manual: {prediksi_manual}")
```

## Tahap 3 — Implementasi dengan scikit-learn (15 menit)

`CategoricalNB` di sklearn membutuhkan input numerik, jadi fitur kategorikal kita ubah dulu dengan `OrdinalEncoder`.

```python
features = ["Outlook", "Temperature", "Humidity", "Wind"]

encoder = OrdinalEncoder()
X = encoder.fit_transform(df[features])
y = df["Play"].values

print("Pemetaan kategori -> angka:")
for col, cats in zip(features, encoder.categories_):
    print(f"  {col}: {dict(enumerate(cats))}")
```

```python
# TODO 7: Latih CategoricalNB (alpha default = 1, pakai Laplace smoothing)
model = CategoricalNB()
model.fit(X, y)

# TODO 8: Hitung akurasi pada data latih (perlu .predict(X) dulu)
y_pred = model.predict(X)
akurasi = (y_pred == y).mean()

print(f"Akurasi pada data latih: {akurasi:.4f}")
```

```python
# TODO 9: Prediksi kasus uji yang SAMA dengan Tahap 2
x_test_df = pd.DataFrame([test])
x_test = encoder.transform(x_test_df[features])

prediksi_sklearn = model.predict(x_test)       # menghasilkan label (0 atau 1)
proba_sklearn = model.predict_proba(x_test)    # menghasilkan probabilitas [P(No), P(Yes)]

print(f"Prediksi sklearn : {prediksi_sklearn[0]}")
print(f"Kelas-kelas      : {model.classes_}")
print(f"Probabilitas     : {proba_sklearn[0]}")
```

**Jawaban Kelompok:**

### 1. Perbandingan Prediksi Manual vs Scikit-learn
* **Apakah prediksi kelas manual sama dengan sklearn?**
    Secara umum **sama**. Meskipun nilai probabilitasnya berbeda, kelas dengan probabilitas tertinggi biasanya tetap konsisten antara hitungan manual dan library.
* **Apakah nilai probabilitas-nya juga sama?**
    **Berbeda.**
* **Jika berbeda, kira-kira kenapa?**
    Perbedaan ini muncul karena parameter **`alpha`** (Laplace/additive smoothing) pada `CategoricalNB`. Secara default, sklearn menggunakan `alpha=1.0`.
    * Hitungan manual seringkali hanya menggunakan frekuensi dasar: P(x|y) = count(x,y) / count(y).
    * Sklearn menggunakan rumus smoothing: P(x|y) = (count(x,y) + alpha) / (count(y) + alpha * n_categories). Hal ini dilakukan untuk menghindari nilai nol.

### 2. Eksperimen `alpha=1e-10`
* **Apakah probabilitas sklearn jadi lebih dekat ke hasil manual?**
    **Ya.** Dengan memberikan nilai `alpha` yang sangat kecil (mendekati nol), efek smoothing hampir hilang. Hasilnya, probabilitas yang dihasilkan oleh Scikit-learn akan menjadi sangat mirip atau identik dengan perhitungan manual yang menggunakan frekuensi murni.

### 3. Kasus Atribut yang Hilang (Zero Frequency)
* **Apa yang terjadi jika kombinasi (kategori, kelas) tidak pernah muncul?**
    Tanpa smoothing, probabilitas untuk kombinasi tersebut adalah **0**. Karena Naive Bayes mengalikan semua probabilitas fitur, satu saja nilai nol akan menyebabkan hasil akhir probabilitas kelas tersebut menjadi **nol mutlak**, meskipun fitur lainnya sangat mendukung.
* **Bagaimana Laplace smoothing menyelamatkan situasi ini?**
    Laplace smoothing menambahkan nilai kecil (`alpha`) ke setiap perhitungan probabilitas. Ini memastikan bahwa tidak ada probabilitas yang bernilai tepat nol. Dengan demikian, model tetap bisa memberikan prediksi yang masuk akal bahkan jika bertemu kombinasi data yang belum pernah dilihat di data latih.

### 4. Asumsi Independensi: Humidity & Outlook
* **Apakah benar-benar independen di dunia nyata?**
    **Tidak.** Di dunia nyata, kelembapan (*Humidity*) seringkali berkaitan dengan cuaca (*Outlook*). Misalnya, saat mendung (*Overcast*) atau hujan (*Rain*), kelembapan biasanya lebih tinggi dibanding saat cerah (*Sunny*).
* **Apa konsekuensinya untuk akurasi model?**
    Karena Naive Bayes menganggap mereka independen, model mungkin akan melakukan "double counting" pada informasi yang berkaitan tersebut. Konsekuensinya, model cenderung menjadi **overconfident** (memberikan probabilitas yang terlalu ekstrem mendekati 0 atau 1). Jika ketergantungan antar fitur sangat kuat, akurasi prediksi bisa menurun karena model gagal menangkap struktur hubungan antar variabel yang sebenarnya.
